# RAG - (Retrieval-Augmented Generation)

**RAG (Retrieval-Augmented Generation)** is one of the most useful concepts to understand if you're getting into modern AI applications.

## 1. What problem does RAG solve?

Imagine you ask an LLM:

> “What is our company's vacation policy?”

A normal LLM only knows what was included in its training or what you put into the current prompt. It **doesn't automatically know your company's private documents**.

RAG solves this by giving the LLM relevant information **at the time you ask the question**.

Think of it as:

**Your question → Find relevant information → Give it to the LLM → Generate answer**

---

## 2. RAG in one simple example

Suppose you have 1,000 company documents:

* HR handbook
* Engineering documentation
* Product manuals
* Company policies
* Meeting notes

A user asks:

> “How many days of parental leave do employees get?”

Instead of asking the LLM to magically know the answer, RAG does this:

### Step 1 — User asks a question

> “How many days of parental leave do employees get?”

### Step 2 — Search your documents

The system searches the company documents for information related to:

> parental leave

It might find:

> “Eligible employees receive 16 weeks of parental leave.”

### Step 3 — Give the result to the LLM

The LLM receives something like:

```text
Context:
Eligible employees receive 16 weeks of parental leave.

Question:
How many days of parental leave do employees get?
```

### Step 4 — LLM generates the answer

> “Eligible employees receive 16 weeks of parental leave.”

That's RAG.

---

# 3. The basic RAG architecture

A typical RAG system looks like this:

```text
                 YOUR DOCUMENTS
                       │
                       ▼
              ┌─────────────────┐
              │  Split into     │
              │     chunks      │
              └────────┬────────┘
                       │
                       ▼
              ┌─────────────────┐
              │   Embeddings    │
              └────────┬────────┘
                       │
                       ▼
              ┌─────────────────┐
              │ Vector Database │
              └────────┬────────┘
                       │
                       │
User Question ─────────┘
       │
       ▼
   Embedding
       │
       ▼
 Search similar chunks
       │
       ▼
 Relevant context
       │
       ▼
      LLM
       │
       ▼
    Answer
```

There are several important pieces here.

---

# 4. Documents

First, you need some knowledge source.

For example:

```text
company_handbook.pdf
product_manual.pdf
engineering_docs.md
faq.html
```

These documents contain the information you want your AI application to use.

---

# 5. Chunking

You usually **don't put an entire 500-page PDF into the LLM**.

Instead, you split it into smaller pieces called **chunks**.

For example:

```text
Document:
Employee Handbook

Chunk 1:
Working hours are from 9 AM to 6 PM...

Chunk 2:
Employees receive 20 days of annual leave...

Chunk 3:
Parental leave is available for...

Chunk 4:
Employees must submit leave requests...
```

Why?

Because when someone asks:

> “What is the parental leave policy?”

you only want to retrieve the relevant section, rather than the entire handbook.

---

# 6. Embeddings — the key concept

This is probably the most important new concept for beginners.

An **embedding** converts text into a list of numbers representing its meaning.

For example, conceptually:

```text
"How much parental leave do I get?"
            ↓
[0.21, -0.14, 0.87, 0.32, ...]
```

And:

```text
"Employees can take 16 weeks of parental leave."
            ↓
[0.19, -0.12, 0.84, 0.35, ...]
```

These vectors are mathematically close because their meanings are similar.

This allows us to perform **semantic search**.

---

# 7. Vector database

The embeddings of your document chunks are stored in a **vector database**.

Examples include:

* Pinecone
* Weaviate
* Milvus
* Qdrant
* pgvector/PostgreSQL
* Chroma

Conceptually:

```text
Chunk                         Embedding

"Annual leave is 20 days"     [0.12, 0.44, ...]
"Parental leave is 16 weeks"  [0.19, 0.83, ...]
"Office hours are 9-6"        [0.72, 0.11, ...]
```

When the user asks a question, you create an embedding for the question and search for nearby vectors.

---

# 8. Retrieval

Suppose the user asks:

> “How long is parental leave?”

The question gets converted into an embedding.

The vector database searches for the most semantically similar chunks.

It might return:

```text
1. "Employees receive 16 weeks of parental leave."
2. "Parental leave is available to eligible employees..."
3. "Employees must notify HR before taking parental leave..."
```

This is the **retrieval** part of RAG.

---

# 9. Augmentation

Now we have:

```text
Question:
How long is parental leave?

Retrieved context:
Employees receive 16 weeks of parental leave.
Parental leave is available to eligible employees.
```

We combine the question and retrieved context into a prompt for the LLM.

That's the **Augmented** part.

---

# 10. Generation

Finally, the LLM reads the context and generates the answer:

> “Eligible employees receive 16 weeks of parental leave.”

That's the **Generation** part.

So:

**R** = Retrieval
**A** = Augmented
**G** = Generation

Hence:

> **Retrieval-Augmented Generation**

---

# 11. RAG vs normal LLM

### Normal LLM

```text
Question
   ↓
LLM
   ↓
Answer
```

### RAG

```text
Question
   ↓
Search knowledge base
   ↓
Relevant information
   ↓
LLM
   ↓
Answer
```

The second approach allows the application to use **external/private/up-to-date information**.

---

# 12. Why not just put everything into the prompt?

You might wonder:

> “Why not send all my documents to the LLM?”

There are several problems:

**1. Context limits**

You may have millions of words but only need 500 relevant words.

**2. Cost**

Sending huge amounts of text repeatedly can be expensive.

**3. Noise**

Giving the model irrelevant information can actually make answers worse.

**4. Freshness**

With RAG, you can update your knowledge base without retraining the model.

---

# 13. RAG does NOT train the LLM

This is a very common beginner misconception.

Suppose you have:

```text
company_policy.pdf
```

You add it to your RAG system.

You **have not trained the LLM on that PDF**.

Instead, you're effectively saying:

> “When someone asks a question, find relevant parts of this document and show them to the LLM.”

That's a major distinction.

### RAG

**Retrieve information → put it in prompt → generate**

### Fine-tuning

**Train/update model behavior using training examples**

They solve different problems.

---

# 14. A simple real-world analogy

Imagine you're taking an exam.

### Without RAG

The teacher asks:

> “What does chapter 7 say?”

You have to answer entirely from memory.

### With RAG

You have a librarian.

You ask:

> “Where does the book discuss parental leave?”

The librarian:

1. Searches the book.
2. Finds the relevant page.
3. Gives it to you.
4. You read it.
5. You answer the question.

The **librarian = retrieval system**.

The **book = knowledge base**.

You **= LLM**.

# 15. The RAG pipeline you should remember

For beginner-level understanding, memorize this:

```text
                 INDEXING

Documents
   ↓
Chunking
   ↓
Embeddings
   ↓
Vector Database


                 QUERYING

User Question
   ↓
Question Embedding
   ↓
Vector Search
   ↓
Top Relevant Chunks
   ↓
Prompt + Context
   ↓
LLM
   ↓
Answer
```

There are really **two phases**:

### Phase 1: Indexing

Prepare your knowledge.

```text
Documents → chunks → embeddings → database
```

### Phase 2: Retrieval/generation

Answer questions.

```text
Question → search → relevant chunks → LLM → answer
```

---

# 16. Where beginners usually get confused

There are three concepts worth keeping separate:

### LLM

Generates text and reasons over the information it receives.

Examples: GPT models, Claude, Gemini, etc.

### Embedding model

Converts text into vectors for semantic comparison.

```text
text → vector
```

### Vector database

Stores vectors and helps find similar vectors.

```text
vector → similar vectors
```

A RAG system commonly combines all three.

---

# 17. What makes a RAG system good?

A basic RAG system is relatively easy to build.

A **good** RAG system is harder.

Important factors include:

### Chunking

Bad chunks → bad retrieval.

### Retrieval

If you retrieve the wrong information, the LLM can't magically fix it.

### Number of chunks

Too few → missing information.

Too many → unnecessary/noisy context.

### Reranking

You can retrieve 20 candidates and then use another model to identify the best 5.

### Metadata filtering

For example:

```text
department = "HR"
year = 2026
document_type = "policy"
```

This can dramatically improve retrieval.

### Prompting

You can instruct the LLM:

> Answer only using the provided context. If the context doesn't contain the answer, say you don't know.

This can reduce unsupported answers.

````python
import os
import numpy as np
import faiss

from groq import Groq
from sentence_transformers import SentenceTransformer
from pypdf import PdfReader
````

### Set-up of Groq - API

````python
from getpass import getpass
GROQ_API_KEY = getpass("Enter your Groq API Key: ")
````

````python
client = Groq(
    api_key=GROQ_API_KEY
)

MODEL = "openai/gpt-oss-20b"
````

### Ask LLM Without RAG

````python
messages = [
    {
        "role": "user",
        "content": "what is inside my private PDF?"
    }
] 

response = client.chat.completions.create(
    model=MODEL,
    messages=messages
)

print(response.choices[0].message.content)
````

### Load Embedding Model

````python
embedding_model = SentenceTransformer('all-MiniLM-L6-v2')
````

### Create Sample Documents

````python
documents = [
    "RAG stands for Retrieval Augmented Generation.",
    "Embedding converts text into vectors.",
    "FAISS is used for similarity search.",
    "Chunking splits large documents into smaller pieces."
]
print(documents)
````

### Generate Embeddings 

````python
embeddings = embedding_model.encode(documents)

print(embeddings.shape)
print(embeddings[0])
````

### Create FAISS Index
(Facebook AI Similarity Search)

````python
# NAIVE APPROACH

# dimension = embeddings.shape[1]

# index = faiss.IndexFlatL2(dimension) # brute force appraoch (Naive Approach)

# index.add(np.array(embeddings, dtype='float32'))

# print("Total vectors stored: ", index.ntotal)
````

````python
# FAISS APPROACH

dimension = embeddings.shape[1]

nlist = 2 # create a no. of clusters

quantizer = faiss.IndexFlatL2(dimension) # quantizer tells the neartest cluster for the query vector
index = faiss.IndexIVFFlat(
    quantizer,
    dimension,
    nlist,
    faiss.METRIC_L2
)
index.train(embeddings)
index.add(embeddings)
````

````python
# Question 
query = "what does RAG stand for? "
query_embedding = embedding_model.encode([query])
````

### Search similar chunks

````python
distances, indices = index.search(
    np.array(query_embedding, dtype="float32"),
    2
)
print(indices)
````

### Retrieve Matching Documents

````python
retrieved_docs = [documents[i] for i in indices[0]]
print(retrieved_docs)
````

### Create Context

````python
context = "\n".join(retrieved_docs)
print(context)
````

### Send Context to LLM

````python
messages = [
    {
        'role': 'system',
        'content': "Answer only using the provided context."
    },
    {
        'role': 'user',
        'content': f"context:\n{context}\n\nquestion: {query}"
    }
]

response = client.chat.completions.create(
    model = MODEL,
    messages = messages
)

print(response.choices[0].message.content)
````

### Load PDF File

````python
pdf = PdfReader("Sample.pdf")

text = ""

for page in pdf.pages:
    text += page.extract_text()

print(text[:1000])
````

### Chunk the PDF text

````python
chunk_size = 300

chunks = []

for i in range(0, len(text), chunk_size):
    chunk = text[i:i + chunk_size]
    chunks.append(chunk)

print("total number of chunks: ", len(chunks))
````

### Generate chunck embeddings

````python
chunk_embeddings = embedding_model.encode(chunks)
````

### Store Chunks in FAISS

````python
# we have a small data so use a naive approach

dimension = chunk_embeddings.shape[1]

pdf_index = faiss.IndexFlatL2(dimension) # brute force appraoch (Naive Approach)

pdf_index.add(np.array(chunk_embeddings, dtype='float32'))

print("Total vectors stored: ", pdf_index.ntotal)
````

````python
# Asking questions on pdf

query = "what is the document about?"
query_embedding = embedding_model.encode([query ])
````

### Retrieve Relevant PDF Chunks

````python
distances, indices = pdf_index.search(
    np.array(query_embedding, dtype="float32"),
    5
)
retrieved_chunks = [chunks[i] for i in indices[0]]
print(retrieved_chunks)
````

````python
context = "\n".join(retrieved_chunks)

messages = [
    {
        'role': 'system',
        'content': "Answer only using the provided context."
    },
    {
        'role': 'user',
        'content': f"context:\n{context}\n\nquestion: {query}"
    }
]

response = client.chat.completions.create(
    model = MODEL,
    messages = messages
)

print(response.choices[0].message.content)
````

# Gradient Descent 🚶‍♂️⛰️

Imagine you're standing **on top of a mountain at night**. 🌙

You want to reach the **lowest point (the valley)**, but because it's dark, you can't see the whole mountain.

What do you do?

* You feel the ground around you.
* Take **one small step downhill**.
* Again check which direction goes downhill.
* Take another small step.
* Repeat this until you reach the bottom.

This is exactly how **Gradient Descent** works in Machine Learning.

---

## What is Gradient Descent?

**Gradient Descent is an optimization algorithm used to find the minimum value of a function by moving step by step in the direction where the function decreases the fastest.**

In Machine Learning:

* The **function** = Cost (Loss) Function
* The **lowest point** = Best model
* The **steps** = Updating model parameters (weights)

---

## Why do we need Gradient Descent?

Suppose we build a Linear Regression model.

The model predicts:

[
y = mx + c
]

Initially, the computer doesn't know the correct values of **m (slope)** and **c (intercept)**.

So it starts with random values.

Example:

```
Actual Line:
y = 2x + 3

Computer starts with:
y = 7x - 5
```

Obviously this prediction is wrong.

Now we need a way to improve these values.

That's where Gradient Descent comes in.

---

## Step-by-Step Process

### Step 1: Initialize Random Parameters

Start with random values.

```
Weight (m) = 8
Bias (c) = -2
```

---

### Step 2: Predict Output

Use these values to predict output.

```
Predicted = mx + c
```

---

### Step 3: Calculate Error

Compare

```
Actual Value
vs
Predicted Value
```

If prediction is wrong

Error becomes large.

---

### Step 4: Calculate Cost Function

The error is converted into one number called **Cost** (or Loss).

Example:

```
Current Cost = 250
```

Our goal is

```
Cost = 0
or
as small as possible.
```

---

### Step 5: Find Gradient

The **gradient** tells us:

> **Which direction should we move to reduce the error?**

Think of it like this:

```
Gradient > 0

Move Left

Gradient < 0

Move Right
```

It also tells **how steep the slope is**.

---

### Step 6: Update Parameters

Weights are updated using

[
\text{New Weight} = \text{Old Weight} - \text{Learning Rate} \times \text{Gradient}
]

Here:

* Old Weight = Current value
* Gradient = Direction
* Learning Rate = Step size

---

### Step 7: Repeat

Again

Predict

↓

Calculate Error

↓

Calculate Cost

↓

Find Gradient

↓

Update Weight

↓

Repeat

Eventually,

Cost becomes very small.

# What is Learning Rate?

The **Learning Rate** decides **how big each step should be**.

### Small Learning Rate

```
0.001
```

```
🙂 Safe
🐢 Very Slow
```

---

### Medium Learning Rate

```
0.01
```

```
🙂 Usually Best
```

---

### Large Learning Rate

```
1.5
```

```
🏃 Big Jumps
May overshoot the minimum.
```

Example:

```
Valley

      O
     /
    /
   /
  X

Small Steps

X → → → → O
```

Large learning rate:

```
X -------->

<--------

-------->

Keeps jumping over the valley.
```

# Why is it called "Gradient" Descent?

* **Gradient** = Slope of the curve.
* **Descent** = Moving downward.

Together:

> "Move in the direction where the slope goes downward the fastest."

---

# Where is Gradient Descent Used?

It is used in almost every Machine Learning algorithm that learns from data, including:

* Linear Regression
* Logistic Regression
* Neural Networks
* Deep Learning
* Linear Models with many parameters

---

# Easy Way to Remember

Gradient Descent works: it measures the error, makes a small adjustment, and repeats until the model performs as well as possible.

### Summary

* **Gradient Descent** helps a model learn by reducing its prediction error.
* It starts with **random weights**.
* It calculates the **error (cost)**.
* The **gradient** shows which direction reduces the error.
* The **learning rate** decides how big the update should be.
* It repeats this process until the **cost is minimized** and the model makes accurate predictions.


````python
from sklearn.datasets import make_regression
import numpy as np
````

````python
X,y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1, noise=80, random_state=13)
````

````python
import matplotlib.pyplot as plt
plt.scatter(X,y)
````

````python
# Lets apply LR (OLS)
from sklearn.linear_model import LinearRegression

reg = LinearRegression()
reg.fit(X, y)
````

````python
reg.coef_
````

````python
reg.intercept_
````

````python
plt.scatter(X,y)
plt.plot(X, reg.predict(X), color='red')
````

Let's apply Gradient Descent assuming slope is constant m = 78.35

and Let's assume the starting value for intercept b =0

````python
y_pred = ((78.35 * X) + 0).reshape(4)
````

````python
plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred,color='#00a65a',label='b = 0')
plt.legend()
plt.show()
````

````python
m = 78.35
b = 100

loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
````

````python
# Lets take learning rate - 0.1

lr = 0.1
step_size = loss_slope*lr
step_size
````

````python
# Calculating the new intercept

b = b - step_size
b
````

````python
# Keep updating the model parameters in a loop until the fitted line converges to the optimal intercept.

y_pred1 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred1,color='#00a65a',label='b = {}'.format(b))
plt.plot(X,y_pred,color='#A3E4D7',label='b = 0')
plt.legend()
plt.show()
````

````python
# Iteration 2 
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
````

````python
step_size = loss_slope*lr
step_size
````

````python
b = b - step_size
b
````

````python
y_pred2 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,reg.predict(X),color='red',label='OLS')
plt.plot(X,y_pred2,color='#00a65a',label='b = {}'.format(b))
plt.plot(X,y_pred1,color='#A3E4D7',label='b = {}'.format(b))
plt.plot(X,y_pred,color='#A3E4D7',label='b = 0')
plt.legend()
plt.show()
````

````python
b = -100
m = 78.35
lr = 0.1

epochs = 20

for i in range(epochs):
  loss_slope = -2 * np.sum(y - m*X.ravel() - b)
  b = b - (lr * loss_slope)

  y_pred = m * X + b

  plt.plot(X,y_pred)

plt.scatter(X,y)
````

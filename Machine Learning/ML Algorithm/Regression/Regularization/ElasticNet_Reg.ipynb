# ELASTICNET REGRESSION

Elastic Net Regression is a machine learning technique used to predict a numerical value (like house price, salary, or sales) while also helping the model decide which input features are important.

It is basically a combination of two methods:

Ridge Regression → uses L2 regularization
Lasso Regression → uses L1 regularization

Elastic Net combines both.

The main formula

Elastic Net adds two penalties to ordinary linear regression: 
### Loss = Prediction Error + λ1​∑∣wi​∣+ λ2​∑(wi)^2​
Where:
- Prediction Error → how wrong the predictions are
- L1 penalty ∑∣wi∣ → comes from Lasso
- L2 penalty (wi)^2​ → comes from Ridge
- λ (lambda) → controls how strongly we penalize large coefficients

```python
from sklearn.datasets import load_diabetes
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression, Ridge,Lasso, ElasticNet
from sklearn.model_selection import train_test_split
```

```python
X,y = load_diabetes(return_X_y=True)

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train, y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Ridge
rid = Ridge(alpha=0.1)
rid.fit(X_train, y_train)
y_pred = rid.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# Lasso
las = Lasso(alpha=0.01)
las.fit(X_train, y_train)
y_pred = las.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# ElasticNet 

el = ElasticNet(alpha=0.005, l1_ratio=0.9)
el.fit(X_train, y_train)
y_pred = el.predict(X_test)
r2_score(y_test,y_pred)
```

# Linear Regression Assumption: 
1. **Linear Relationship (between independent and dependent variables)**

   * The relationship between the predictors (X) and the response variable (Y) should be linear.
   * **Why?** Linear regression models a straight-line relationship.
   * **How to check?** Scatter plots, residual vs. fitted plots.

2. **No Multicollinearity**

   * Independent variables should not be highly correlated with each other.
   * **Why?** High multicollinearity makes coefficient estimates unstable and difficult to interpret.
   * **How to check?** Variance Inflation Factor (VIF). Generally:

     * VIF < 5 → Acceptable
     * VIF > 10 → Serious multicollinearity

3. **Normality of Residuals**

   * The residuals (errors) should be approximately normally distributed.
   * **Why?** Required for valid hypothesis tests and confidence intervals.
   * **How to check?**

     * Histogram of residuals
     * Q-Q plot
     * Shapiro-Wilk test

4. **Homoscedasticity (Constant Variance of Errors)**

   * The variance of residuals should remain constant across all levels of the independent variables.
   * **Why?** Unequal variance (heteroscedasticity) leads to inefficient estimates and unreliable statistical tests.
   * **How to check?**

     * Residual vs. fitted plot
     * Breusch-Pagan test
     * White test

5. **No Autocorrelation of Errors**

   * Residuals should be independent of one another.
   * **Why?** This assumption is especially important for time-series data.
   * **How to check?**

     * Durbin-Watson test

       * ≈ 2 → No autocorrelation
       * < 2 → Positive autocorrelation
       * > 2 → Negative autocorrelation

### Quick Summary

| Assumption             | Purpose                                | Common Test                       |
| ---------------------- | -------------------------------------- | --------------------------------- |
| Linear relationship    | X and Y have a linear relationship     | Scatter plot, residual plot       |
| No multicollinearity   | Predictors are not highly correlated   | VIF                               |
| Normality of residuals | Residuals follow a normal distribution | Q-Q plot, Shapiro-Wilk test       |
| Homoscedasticity       | Constant variance of residuals         | Residual plot, Breusch-Pagan test |
| No autocorrelation     | Residuals are independent              | Durbin-Watson test                |

These are the standard assumptions typically expected in interviews, exams, and practical machine learning/statistics applications.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('data.csv')
df.head()
```

```python
X = df.iloc[:,0:3].values
y = df.iloc[:,-1].values
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=1)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
# Residual
y_pred = model.predict(X_test)
residual = y_test - y_pred
```

### 1. Linear Relationship 

```python
fig, (ax1, ax2, ax3) = plt.subplots(ncols=3, figsize=(12, 4.5))

ax1.scatter(df['feature1'], df['target'])
ax1.set_title("Feature1")
ax2.scatter(df['feature2'], df['target'])
ax2.set_title("Feature2")
ax3.scatter(df['feature3'], df['target'])
ax3.set_title("Feature3")

plt.show()
```

### 2. Multicollinearity

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif = []

for i in range(X_train.shape[1]):
    vif.append(variance_inflation_factor(X_train, i))
```

```python
pd.DataFrame({'vif': vif}, index=df.columns[0:3]).T
```

```python
# Another Technique
sns.heatmap(df.iloc[:,0:3].corr(), annot=True)
```

### 3. Normal Residual

```python
sns.displot(residual, kind='kde')
```
